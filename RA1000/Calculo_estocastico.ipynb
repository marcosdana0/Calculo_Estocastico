Para el lanzamiento de dos dados, cada dado tiene 6 posibles resultados (1, 2, 3, 4, 5, 6). El espacio muestral consiste en todas las combinaciones posibles de los resultados de ambos dados.

$$ P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)} $$Donde:
*   $P(A|B)$ es la probabilidad de que el evento A ocurra dado que el evento B ha ocurrido (probabilidad a posteriori).
*   $P(B|A)$ es la probabilidad de que el evento B ocurra dado que el evento A ha ocurrido.
*   $P(A)$ es la probabilidad a priori de que el evento A ocurra.
*   $P(B)$ es la probabilidad a priori de que el evento B ocurra.
